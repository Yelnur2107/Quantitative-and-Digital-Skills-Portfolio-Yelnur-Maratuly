```python
import pandas as pd
```

```python
df = pd.read_csv(r"C:\Users\huawei\Downloads\data-final.csv", sep="\t")
```

```python
df
```

```python
cols = [col for col in df.columns 
        if col.startswith(("EXT","EST","AGR","CSN","OPN")) 
        and not col.endswith("_E")]

df_questions = df[cols]

df_questions.shape
df_questions.head()
```

```python
df_sample = df_questions.sample(n=2000, random_state=42)
df_sample
```

```python
from factor_analyzer import FactorAnalyzer

fa = FactorAnalyzer(rotation=None)
fa.fit(df_sample)

ev, v = fa.get_eigenvalues()
```

```python
ev
```

```python
fa = FactorAnalyzer(n_factors=5, rotation='varimax')
fa.fit(df_sample)
```

```python
import pandas as pd

loadings = pd.DataFrame(
    fa.loadings_,
    index=df_sample.columns,
    columns=[f'Factor{i+1}' for i in range(5)]
)

loadings.head(100)
```
